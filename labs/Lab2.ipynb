## **Lab 2: Introduction to Complex Exponentials**

The goal of this lab is to get comfortable with complex numbers and to see how they help us describe sinusoidal signals.

A real sinusoid can be written as

$$x(t)=\Re\{A e^{j(2\pi f_0 t+\phi)}\}=A\cos(2\pi f_0 t+\phi).$$

This is the main idea for this lab: a complex exponential lets us keep amplitude, frequency, and phase together in one compact expression. When we take the real part, we recover the sinusoid we can plot and hear.

**Complex Numbers in Python**

Python makes it easy to work with complex numbers and to visualize them as points or arrows in the complex plane.

*conj()*       Complex conjugate

*x = 3 + 4j*    the `j` suffix defines the imaginary unit

exp(1j*theta)  the complex exponential

### 2. Complex Sinusoids

Now let’s work with complex sinusoids. They are useful because the angle of the complex number tells us the phase, and its rotation in the complex plane corresponds to the oscillation over time.

A simple example is:

$$x(t)=A e^{j(2\pi f_0 t+\phi)}.$$

In this lab, we will use the real part of this signal to get the sinusoid we can plot and hear.

````python
A = 1
f0 = 440
fs = 44100
phi = np.pi/2

t = np.arange(0, 0.01, 1.0/fs)
x = A * np.exp(1j*(2*np.pi*f0*t + phi))
````

Now we can plot the real and imaginary parts of this signal. The real part looks like a cosine-like waveform, while the imaginary part looks like a sine-like waveform.

Try changing the phase $\phi$ and see how the waveform shifts.

````python
plot_signals([np.real(x), np.imag(x)], fs, name=['real part', 'imag part'])
````

3.2. Now, define a harmonic signal, `y` whose fundamental frequency is the one defined in Lab 1. Go step by step adding a new harmonic in each step. Plot both signals (the reference and the synthesized) and try to reproduce the shape of the reference signal. Play both sounds and compare them carefully.

**Note 1**: in order to have a similar shape, we need to select the amplitudes and phases carefully. One way to do this is to define the harmonic signal as follows:

$$y(t) = \sum_{k=1}^K A_k\cos\left(2\pi kf_0 t + k \phi - (k-1)\pi/2 \right), $$

where $K$ is the number of harmonics, $f_0$ is the fundamental frequency, $A_k$ is the amplitude (weight) of each harmonic and $\phi$ is the phase of the signal (defined in Lab1 Ex 2.4).

**Note 2**:
Define the $A_k$ values relative to the fundamental frequency. This means to define $A_1=1$ and the others less than 1. You can use Audacity to plot the spectrum of the fragment selected of the reference audio and measure the relative amplitudes of the harmonics.

**Note 3**: Normalize the amplitude of the signal by the same amplitude of the reference. For instance, if the amplitude of the reference signal is 0.33, you can normalize the syntesized signal by first dividing by its maximum and then multiplyng by 0.33:

```
y = 0.33 * y / np.amax(y)
```

````python
import matplotlib.pyplot as plt

K = 5 # number of harmonics
f0 = 373.73 # fundamental frequency
phi = 3.14 # initial phase

t = np.arange(len(ref)) / fs # time array for the duration

A_k = [1.0, 0.5, 0.25, 0.1, 0.05] # relative amplitudes to the fundamental frequency

y = np.zeros_like(t, dtype=float)
for k in range(1, K + 1):
    amp = A_k[k-1] # Get the amplitude for the current harmonic
    phase_k = (k * phi) - ((k - 1) * (np.pi / 2)) # calculate the phase for the current harmonic
    
    # Add the current harmonic to the total signal
    y += amp * np.cos(2 * np.pi * k * f0 * t + phase_k)

# we find the higher value of the real harp harmonic
max_ref_amp = np.amax(np.abs(ref))
y = max_ref_amp * (y / np.amax(np.abs(y)))

period = 0.002675 
t_start = 0.2 # we start at 0.2 seconds, as in the lab 1
t_end = t_start + (8 * period) # the end time is 8 periods after the start time

start_idx = int(t_start * fs)
end_idx = int(t_end * fs)

plt.figure(figsize=(12, 4))
plt.plot(t[start_idx:end_idx], ref[start_idx:end_idx], label='Reference (Harp)')
plt.plot(t[start_idx:end_idx], y[start_idx:end_idx], label=f'Harmonic Synthesis (K={K})')
plt.xlabel('Time (s)')
plt.ylabel('Amplitude')
plt.title('Comparison: Reference vs Harmonic Synthesis')
plt.legend()
plt.show()

ipd.Audio(y, rate=fs)
````
